```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('B5_Stores1b.csv')
df.head()
```

```python
#2
missing_percentage = df.isnull().mean() * 100
print(missing_percentage)
```

```python
#3
temp_df = df.copy()

for column in temp_df.columns:
    if temp_df[column].isnull().mean() * 100 < 10:
        if temp_df[column].dtype in ['float64', 'int64']: 
            temp_df[column].fillna(temp_df[column].mean(), inplace=True)
print(temp_df)
```

```python
#4
temp_df = df.copy()

for column in temp_df.columns:
    if temp_df[column].isnull().mean() * 100 < 10:
        if temp_df[column].dtype in ['float64', 'int64']: 
            temp_df[column].interpolate(method='nearest', inplace=True)
print(temp_df)
```

```python
#5
for column in temp_df.columns:
    if temp_df[column].isnull().mean() * 100 < 10:
        if temp_df[column].dtype == 'object':  
            temp_df[column].fillna(temp_df[column].mode()[0], inplace=True)

print(temp_df)
```

```python
#6
from scipy import stats
temp_df = df.copy()

temp_df.dropna(thresh=len(temp_df) * 0.9, axis=1, inplace=True)
print(temp_df.shape)
```

```python
#7
z_scores = stats.zscore(temp_df['Quantity'])
temp_df = temp_df[(z_scores < 3)]

print(temp_df.shape)
```

```python
columns_with_nulls = temp_df.columns[temp_df.isnull().any()]

print(columns_with_nulls)
```

```python
#8
temp_df = df.copy()

duplicate_rows_all = temp_df[temp_df.duplicated()]

duplicate_percentage_all = (len(duplicate_rows_all) / len(temp_df)) * 100
print(f"Percentage of duplicate rows: {duplicate_percentage_all:.2f}%")
```

```python
#9
duplicate_rows_specific = temp_df[temp_df.duplicated(subset=['Customer', 'Product line', 'Age', 'Gender'])]
duplicate_percentage_specific = (len(duplicate_rows_specific) / len(temp_df)) * 100
print(f"Percentage of duplicate rows (specific columns): {duplicate_percentage_specific:.2f}%")

temp_df.drop_duplicates(subset=['Customer', 'Product line', 'Age', 'Gender'], inplace=True)

print(temp_df.shape)
```

```python
#10
from sklearn.preprocessing import MinMaxScaler
temp_df = df.copy()

scaler = MinMaxScaler()
temp_df['Age_normalized'] = scaler.fit_transform(temp_df[['Age']])

plt.scatter(temp_df.index, temp_df['Age_normalized'])
plt.title('Min-Max Normalized Age')
plt.xlabel('Index')
plt.ylabel('Normalized Age')
plt.show()
```

```python
#11
from scipy.stats import zscore
temp_df['Age_zscore'] = zscore(temp_df['Age'])

plt.scatter(temp_df.index, temp_df['Age_zscore'])
plt.title('Z-score Normalized Age')
plt.xlabel('Index')
plt.ylabel('Z-score Age')
plt.show()
```

```python
#12
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
temp_df['Payment_encoded'] = label_encoder.fit_transform(temp_df['Payment'])
print(temp_df[['Payment', 'Payment_encoded']].head())
```

```python
#13
from sklearn.preprocessing import OneHotEncoder
one_hot_encoder = OneHotEncoder(sparse_output=False)
payment_encoded = one_hot_encoder.fit_transform(temp_df[['Payment']])
payment_encoded_df = pd.DataFrame(payment_encoded, columns=one_hot_encoder.get_feature_names_out(['Payment']))
temp_df = temp_df.join(payment_encoded_df)
print(temp_df.head())
```
